```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools
# --- PART 1C: Bayes' Theorem ---
P_Disease = 0.01    # Only 1% of the population has the disease.
P_Pos_D   = 0.99    # Sensitivity - If a person has the disease, the probability that the test is positive is 99%.
P_Pos_ND  = 0.05    # False Positive Rate - If a person does NOT have the disease, there is still a 5% probability that the test is positive.


# Total Probability
P_Pos = P_Pos_D*P_Disease + P_Pos_ND*(1-P_Disease)
# Bayes' Theorem
P_D_Pos = (P_Pos_D * P_Disease) / P_Pos
print(f'P(Disease | Positive) = {P_D_Pos:.4f} ({P_D_Pos*100:.2f}%)')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools
# --- VISUALIZATION ---
# Empirical convergence plot
n_total = 10000
tosses  = np.random.binomial(1, 0.5, n_total)
emp_prob = [tosses[:i].mean() for i in range(1, n_total+1)]
plt.figure(figsize=(10,4))
plt.plot(emp_prob, color='steelblue', linewidth=1)
plt.axhline(0.5, color='red', linestyle='--', label='True P=0.5')
plt.xscale('log')
plt.title('Empirical Probability Convergence (Law of Large Numbers)')
plt.xlabel('Number of Trials (log scale)')
plt.ylabel('Empirical Probability')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
#Exercise 1
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import itertools

np.random.seed(42)

# --- PART 1A: Classical Probability ---
coin_space = ['H', 'T']
three_coin = list(itertools.product(coin_space, coin_space,coin_space))
print('Sample Space:', three_coin)

three_heads = [s for s in three_coin if s.count('H') == 3]
P_3H = len(three_heads) / len(three_coin)
print(f'P(3 Heads) = {P_3H}')

two_heads = [s for s in three_coin if s.count('H') == 2]
P_2H = len(two_heads) / len(three_coin)
print(f'P(2 Heads) = {P_2H}')

no_head = [ s for s in three_coin if s.count('H') == 0 ]
P_nH = len(no_head) / len(three_coin)
print(f'P(atleast 1 Head) = {1-P_nH}')
```

```python
#Exercise 3
import numpy as np

# Sample space of two dice
dice1 = np.array([1, 2, 3, 4, 5, 6])
dice2 = np.array([1, 2, 3, 4, 5, 6])

sample_space = np.array(np.meshgrid(dice1, dice2)).T.reshape(-1, 2)

print("Sample Space:")
print(sample_space)

total = len(sample_space)

# a) Sum is 7
a = np.sum(sample_space[:, 0] + sample_space[:, 1] == 7)
print("a) Probability of sum 7:", a / total)

# b) Sum is greater than 8
b = np.sum(sample_space[:, 0] + sample_space[:, 1] > 8)
print("b) Probability of sum greater than 8:", b / total)

# c) Both dice show same number
c = np.sum(sample_space[:, 0] == sample_space[:, 1])
print("c) Probability of same number:", c / total)

# d) At least one die shows 6
d = np.sum((sample_space[:, 0] == 6) | (sample_space[:, 1] == 6))
print("d) Probability of at least one 6:", d / total)
```

```python
#EXERCISE 4
import numpy as np
import matplotlib.pyplot as plt

# 10,000 coin tosses
tosses = np.random.randint(0, 2, 10000)

# Cumulative probability of Heads
heads = np.cumsum(tosses)
toss_number = np.arange(1, 10001)
probability = heads / toss_number

# Plot
plt.plot(toss_number, probability)
plt.axhline(0.5, linestyle='--')

plt.xscale('log')
plt.xlabel("Number of Tosses")
plt.ylabel("Cumulative Probability of Heads")
plt.title("Law of Large Numbers")
plt.show()
```

```python
#Exercise 5
import numpy as np

# Sample space
die = np.array([1, 2, 3, 4, 5, 6])

# A = even number
A = die % 2 == 0

# B = number greater than 3
B = die > 3

# A and B
A_and_B = A & B

# P(A|B) = P(A and B) / P(B)
P_A_given_B = np.sum(A_and_B) / np.sum(B)

print("P(A|B) =", P_A_given_B)
```

```python
#Exercise 6
import numpy as np

# Given values
P_D = 0.02          # Probability of disease
P_positive_D = 0.95 # Sensitivity
P_positive_not_D = 0.04  # False-positive rate

# Bayes' Theorem
P_positive = (P_positive_D * P_D) + (P_positive_not_D * (1 - P_D))

P_D_positive = (P_positive_D * P_D) / P_positive

print("P(Disease | Positive) =", P_D_positive * 100, "%")
```

```python
#Exercise 7
import numpy as np

# Given values
P_spam = 0.20
P_offer_spam = 0.80
P_offer_not_spam = 0.10

# Probability of getting "offer"
P_offer = (P_offer_spam * P_spam) + (P_offer_not_spam * (1 - P_spam))

# Bayes theorem
P_spam_offer = (P_offer_spam * P_spam) / P_offer

print("P(Spam | offer) =", P_spam_offer * 100, "%")
```
